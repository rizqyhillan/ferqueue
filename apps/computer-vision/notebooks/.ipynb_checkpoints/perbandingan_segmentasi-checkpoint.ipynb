# FerQueue · Bounding box dengan dan tanpa segmentasi

Bandingkan dua hasil dari **deteksi SSD yang sama** pada gambar test FerQueue.

- **Tanpa segmentasi:** gunakan bounding box asli dari SSD setelah filter confidence dan NMS.
- **Dengan segmentasi:** jalankan GrabCut di dalam setiap kotak SSD. Ganti kotak dengan persegi panjang yang mengelilingi mask. Jika mask kosong atau GrabCut gagal, pakai kotak awal.

Kelas dan confidence tetap berasal dari SSD. Cara ini menguji apakah segmentasi membantu **merapikan kotak yang sudah terdeteksi**. Anotasi test dipakai untuk penilaian dan visualisasi. Eksperimen dijalankan tanpa training ulang.

Pilih kernel Python 3.11 dari environment PCV FerQueue, lalu jalankan **Run All Cells**. Notebook mencari root FerQueue dari direktori kerja Jupyter. Kalau dibuka dari luar repo, isi `ROOT_OVERRIDE` dengan lokasi folder FerQueue. Jika environment PCV belum muncul sebagai kernel, jalankan perintah berikut di terminal dari root repo.

```bash
.venv-pcv/bin/python -m pip install ipykernel
.venv-pcv/bin/python -m ipykernel install --user --name ferqueue-pcv --display-name "Python (FerQueue PCV)"
```

Notebook memakai TensorFlow, OpenCV, Pillow, NumPy, Matplotlib, dan pycocotools di kernel tersebut.

## 1. Pilih model dan jumlah gambar

Default memakai ekspor lokal `pcv-cpu-full-001` beserta split test yang dibekukan saat training. Coba 24 gambar dahulu, lalu set `LIMIT = 0` untuk menilai semua gambar test. Jika memakai model impor, misalnya `pcv-receiver-001`, pilih `receiver.json` dan ganti `RUN_DIR` ke folder model itu. Pastikan gambar test terpisah dari data training model impor.

````python
from pathlib import Path
import random
import sys

ROOT_OVERRIDE = None  # Contoh: Path("/lokasi/ferqueue")
CONFIG_PATH = "apps/computer-vision/configs/coco.json"
RUN_DIR = "models/mobilenet/runs/pcv-cpu-full-001"
LIMIT = 24  # 0 = semua gambar test
PREVIEW_COUNT = 4
SEED = 42

cwd = Path.cwd()
root = Path(ROOT_OVERRIDE).expanduser().resolve() if ROOT_OVERRIDE else next(
    (folder for folder in (cwd, *cwd.parents) if (folder / "pcv.py").is_file()),
    None,
)
if root is None or not (root / "pcv.py").is_file():
    raise FileNotFoundError("Isi ROOT_OVERRIDE dengan lokasi folder FerQueue.")
if LIMIT < 0 or PREVIEW_COUNT < 0:
    raise ValueError("LIMIT dan PREVIEW_COUNT harus >= 0.")

src = str(root / "apps/computer-vision/src")
if src not in sys.path:
    sys.path.insert(0, src)

from ferqueue_cv.dataset import verify_images
from ferqueue_cv.settings import load_config
from ferqueue_cv.training import load_detector, load_manifest

config = load_config(CONFIG_PATH, root)
config["run_dir"] = RUN_DIR
detector, frozen, metadata = load_detector(config, root)
manifest = frozen["manifest"]
if "splits" in manifest:
    verify_images(manifest, root)
else:
    manifest = load_manifest(config, root)
    print("Model impor memakai split test lokal. Periksa pemisahannya dari data training.")

classes = metadata["classes"]
if classes != manifest["classes"]:
    raise ValueError("Nama atau urutan kelas model berbeda dari manifest test.")

test_rows = manifest["splits"]["test"]
rows = random.Random(SEED).sample(test_rows, LIMIT) if 0 < LIMIT < len(test_rows) else test_rows
print(f'{metadata["checkpoint"]} · {len(rows)}/{len(test_rows)} gambar test')
print(f'Confidence ≥ {config["confidence_threshold"]} · IoU NMS {config.get("nms_iou_threshold", 0.45)}')
````

## 2. Buat dua versi bounding box

Kedua versi memakai preprocessing dari metadata model serta pengaturan confidence dan NMS yang sama. Ukuran gambar tetap sama sehingga kedua kotak bisa langsung dibandingkan dengan anotasi.

````python
from collections import Counter
from time import perf_counter

import numpy as np
from PIL import Image, ImageChops

from ferqueue_cv.postprocessing import select_detections
from ferqueue_cv.processing import grabcut_mask, preprocess_image
from ferqueue_cv.training import detect


def baseline_boxes(image):
    result = detect(detector, image)
    width, height = image.size
    boxes = []
    for index in select_detections(result, config):
        class_id = int(result["detection_classes"][index])
        ymin, xmin, ymax, xmax = np.clip(result["detection_boxes"][index], 0, 1)
        if 1 <= class_id <= len(classes) and xmin < xmax and ymin < ymax:
            boxes.append({
                "class_id": class_id,
                "score": float(result["detection_scores"][index]),
                "xyxy": [xmin * width, ymin * height, xmax * width, ymax * height],
            })
    return boxes


def refine_boxes(image, boxes):
    width, height = image.size
    refined = []
    masks = []
    failures = Counter()
    for box in boxes:
        xmin, ymin, xmax, ymax = box["xyxy"]
        seed_box = {
            "xmin": xmin / width, "ymin": ymin / height,
            "xmax": xmax / width, "ymax": ymax / height,
        }
        mask, status = grabcut_mask(image, seed_box, config.get("segmentation"))
        if mask is None:
            failures[status] += 1
            refined.append(box)
        else:
            refined.append({**box, "xyxy": mask.getbbox()})
            masks.append(mask)
    return refined, masks, failures


def coco_bbox(xyxy):
    xmin, ymin, xmax, ymax = map(float, xyxy)
    return [xmin, ymin, xmax - xmin, ymax - ymin]


groundtruth = {
    "images": [],
    "annotations": [],
    "categories": [{"id": i, "name": name} for i, name in enumerate(classes, 1)],
}
predictions = {"Tanpa segmentasi": [], "Dengan GrabCut": []}
previews = []
fallbacks = Counter()
segmentation_ms = []

for image_id, row in enumerate(rows, 1):
    with Image.open(root / row["image"]) as source:
        image = preprocess_image(source, metadata.get("preprocessing"))
    width, height = image.size
    baseline = baseline_boxes(image)

    started = perf_counter()
    segmented, masks, failures = refine_boxes(image, baseline)
    segmentation_ms.append((perf_counter() - started) * 1000)
    fallbacks.update(failures)

    references = [
        {
            "class_id": box["class_id"],
            "xyxy": [box["xmin"] * width, box["ymin"] * height,
                     box["xmax"] * width, box["ymax"] * height],
        }
        for box in row["boxes"]
    ]
    groundtruth["images"].append({
        "id": image_id, "file_name": row["image"], "width": width, "height": height,
    })
    for box in references:
        bbox = coco_bbox(box["xyxy"])
        groundtruth["annotations"].append({
            "id": len(groundtruth["annotations"]) + 1,
            "image_id": image_id, "category_id": box["class_id"], "bbox": bbox,
            "area": bbox[2] * bbox[3], "iscrowd": 0,
        })
    for mode, boxes in zip(predictions, (baseline, segmented)):
        predictions[mode].extend({
            "image_id": image_id, "category_id": box["class_id"],
            "bbox": coco_bbox(box["xyxy"]), "score": box["score"],
        } for box in boxes)

    if image_id <= PREVIEW_COUNT:
        mask_preview = Image.new("L", image.size)
        for mask in masks:
            mask_preview = ImageChops.lighter(mask_preview, mask)
        previews.append((row["image"], image, references, baseline, segmented, mask_preview))
    if image_id % 25 == 0 or image_id == len(rows):
        print(f"{image_id}/{len(rows)} gambar selesai")
````

## 3. Ukur hasil

**IoU** mengukur seberapa besar tumpang tindih antara kotak prediksi dan anotasi, dari 0 sampai 1. **AP50** menerima kecocokan pada IoU ≥ 0,50. **AP75** menuntut kotak yang lebih rapat pada IoU ≥ 0,75. **mAP** merangkum beberapa ambang IoU. Nilai yang lebih tinggi berarti hasil lebih baik pada sampel ini.

Metrik dihitung dari deteksi yang lolos filter FerQueue. Bandingkan juga jumlah kotak yang kembali ke kotak awal saat GrabCut gagal dan tambahan waktu per gambar.

````python
from contextlib import redirect_stdout
from io import StringIO
from statistics import median

from ferqueue_cv.coco_metrics import compute_coco_metrics

if groundtruth["annotations"]:
    with redirect_stdout(StringIO()):
        metrics = {
            mode: compute_coco_metrics(groundtruth, boxes)
            for mode, boxes in predictions.items()
        }

    print(f'{"Metrik":<14} {"Tanpa segmentasi":>17} {"Dengan GrabCut":>16} {"Selisih":>9}')
    for label, key in (
        ("AP50", "Precision/mAP@.50IOU"),
        ("AP75", "Precision/mAP@.75IOU"),
        ("mAP 50–95", "Precision/mAP"),
    ):
        baseline_score, segmented_score = (
            result["DetectionBoxes_" + key] for result in metrics.values()
        )
        print(f"{label:<14} {baseline_score:>17.3f} {segmented_score:>16.3f} "
              f"{segmented_score - baseline_score:>+9.3f}")
else:
    print("Sampel ini tidak memiliki anotasi kendaraan. Tambah LIMIT atau gunakan seluruh split test.")

counts = Counter(box["category_id"] for box in groundtruth["annotations"])
print("\nAnotasi per kelas:", {name: counts[i] for i, name in enumerate(classes, 1)})
total_boxes = len(predictions["Tanpa segmentasi"])
print(f"Mask berhasil dibuat untuk {total_boxes - fallbacks.total()}/{total_boxes} kotak SSD.")
if fallbacks:
    print("Kotak awal dipertahankan:", dict(fallbacks))
print(f"Tambahan waktu GrabCut per gambar (median): {median(segmentation_ms):.1f} ms")
````

## 4. Lihat kotak dan mask

Garis **hijau** menunjukkan prediksi. Garis putus-putus **kuning** menunjukkan anotasi test. Panel kanan menunjukkan piksel yang dipilih GrabCut. Naikkan `PREVIEW_COUNT` untuk melihat lebih banyak contoh.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle


def draw_boxes(ax, boxes, *, color, linestyle="-"):
    for box in boxes:
        xmin, ymin, xmax, ymax = box["xyxy"]
        ax.add_patch(Rectangle(
            (xmin, ymin), xmax - xmin, ymax - ymin,
            fill=False, edgecolor=color, linestyle=linestyle, linewidth=2,
        ))
        if "score" in box:
            name = classes[box["class_id"] - 1]
            ax.text(
                xmin, ymin, f'{name} {box["score"]:.2f}',
                color=color, fontsize=9, va="top",
                bbox={"facecolor": "black", "alpha": 0.7, "pad": 2},
            )


for filename, image, references, baseline, segmented, mask in previews:
    fig, axes = plt.subplots(1, 3, figsize=(14, 4), layout="constrained")
    fig.suptitle(Path(filename).name)
    for ax, title, boxes in zip(axes[:2], predictions, (baseline, segmented)):
        ax.imshow(image, extent=(0, image.width, image.height, 0))
        draw_boxes(ax, references, color="gold", linestyle="--")
        draw_boxes(ax, boxes, color="springgreen")
        ax.set_title(f"{title} · {len(boxes)} kotak")
    axes[2].imshow(mask, cmap="gray", vmin=0, vmax=255)
    axes[2].set_title("Mask GrabCut")
    for ax in axes:
        ax.axis("off")
    plt.show()
````
